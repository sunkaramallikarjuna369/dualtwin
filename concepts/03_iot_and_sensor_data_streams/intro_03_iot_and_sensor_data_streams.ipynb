## What is IoT and Sensor Data Streams?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Think of sensors as the eyes and ears of your digital twin - they constantly watch and listen to what's happening in the physical world.

### Technical Definition

The infrastructure and protocols for collecting, transmitting, and processing real-time data from physical sensors to digital twin systems.

```
# IoT Data Stream Demo
@dataclass
class SensorStream:
    """Simulated sensor data stream."""
    sensor_id: str
    data_type: str
    sample_rate: float = 1.0  # Hz
    
    def generate_reading(self) -> Dict[str, Any]:
        return {
            'sensor_id': self.sensor_id,
            'timestamp': datetime.now(),
            'value': random.gauss(25, 2),
            'quality': random.uniform(0.9, 1.0)
        }

# Create sensor streams
temp_sensor = SensorStream("TEMP-001", "temperature")
pressure_sensor = SensorStream("PRESS-001", "pressure")

# Generate readings
for _ in range(5):
    reading = temp_sensor.generate_reading()
    print(f"Sensor: {reading['sensor_id']}, Value: {reading['value']:.2f}")
```

## Why IoT and Sensor Data Streams Matters

IoT data streams are the lifeblood of digital twins - without real-time data, twins cannot stay synchronized with physical reality.
