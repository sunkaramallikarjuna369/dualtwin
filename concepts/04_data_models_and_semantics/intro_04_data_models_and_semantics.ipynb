## What is Data Models and Semantics?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Just like a dictionary defines words, data models define what information means in your digital twin system.

### Technical Definition

Formal structures that define the schema, relationships, and meaning of data within digital twin systems, often using ontologies and semantic web technologies.

```
# Data Model Demo
@dataclass
class TwinProperty:
    """Semantic property definition."""
    name: str
    data_type: str
    unit: str
    description: str

@dataclass
class TwinModel:
    """Digital twin data model."""
    model_id: str
    properties: List[TwinProperty] = field(default_factory=list)
    
    def validate_data(self, data: Dict) -> bool:
        for prop in self.properties:
            if prop.name not in data:
                return False
        return True

# Define a pump model
pump_model = TwinModel("PumpModel", properties=[
    TwinProperty("temperature", "float", "celsius", "Operating temperature"),
    TwinProperty("pressure", "float", "bar", "Output pressure"),
    TwinProperty("flow_rate", "float", "l/min", "Flow rate"),
])

# Validate data
test_data = {"temperature": 45.0, "pressure": 5.2, "flow_rate": 120.0}
print(f"Data valid: {pump_model.validate_data(test_data)}")
```

## Why Data Models and Semantics Matters

Proper data models ensure that information is consistent, interoperable, and meaningful across the entire digital twin ecosystem.
