# 09 - GenAI with Digital Twins: Examples

This notebook provides practical examples demonstrating genai with digital twins concepts.

```
import sys
sys.path.insert(0, '../../..')

import random
from datetime import datetime, timedelta
from dataclasses import dataclass, field
from typing import Dict, List, Any, Optional
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
```

## Example 1: Basic Implementation

This example demonstrates the fundamental concepts of genai with digital twins.

```
# Example 1: Basic 09 Genai With Digital Twins Implementation
print("Running Example 1...")

# Create sample data
data = {
    'timestamp': datetime.now(),
    'values': [random.gauss(25, 2) for _ in range(20)]
}

# Process data
mean_val = np.mean(data['values'])
std_val = np.std(data['values'])

print(f"Mean: {mean_val:.2f}")
print(f"Std Dev: {std_val:.2f}")

# Visualize
plt.figure(figsize=(10, 4))
plt.plot(data['values'], 'b-o')
plt.axhline(y=mean_val, color='r', linestyle='--', label=f'Mean: {mean_val:.2f}')
plt.fill_between(range(len(data['values'])), mean_val-std_val, mean_val+std_val, alpha=0.2)
plt.legend()
plt.title('Example 1: Data Analysis')
plt.show()
```

## Example 2: Advanced Implementation

This example shows more advanced usage patterns.

```
# Example 2: Advanced 09 Genai With Digital Twins Implementation
print("Running Example 2...")

@dataclass
class AdvancedTwin:
    """Advanced twin implementation."""
    twin_id: str
    data: Dict[str, Any] = field(default_factory=dict)
    history: List[Dict] = field(default_factory=list)
    
    def update(self, new_data: Dict) -> None:
        self.data.update(new_data)
        self.history.append({'timestamp': datetime.now(), **new_data})
    
    def get_statistics(self) -> Dict:
        if not self.history:
            return {}
        df = pd.DataFrame(self.history)
        return df.describe().to_dict()

# Create and use twin
twin = AdvancedTwin("ADV-001")
for i in range(30):
    twin.update({'value': random.gauss(50, 10), 'quality': random.uniform(0.8, 1.0)})

stats = twin.get_statistics()
print(f"Collected {len(twin.history)} data points")
print(f"Value mean: {stats.get('value', {}).get('mean', 0):.2f}")
```
