## What is GenAI with Digital Twins?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Using AI that can generate reports, answer questions, and explore scenarios in natural language.

### Technical Definition

Integration of generative AI capabilities for natural language interfaces, automated reporting, and scenario exploration.

```
# GenAI Integration Demo (Simulated)
class TwinCopilot:
    """Simulated GenAI copilot for digital twins."""
    
    def __init__(self, twin_data: Dict):
        self.twin_data = twin_data
    
    def answer_query(self, query: str) -> str:
        """Simulate answering natural language queries."""
        query_lower = query.lower()
        
        if 'temperature' in query_lower:
            return f"Current temperature is {self.twin_data.get('temperature', 'N/A')}°C"
        elif 'status' in query_lower:
            return f"System status: {self.twin_data.get('status', 'unknown')}"
        elif 'recommendation' in query_lower:
            if self.twin_data.get('temperature', 0) > 40:
                return "Recommendation: Reduce load to prevent overheating"
            return "System operating normally, no action needed"
        return "I can help with temperature, status, and recommendations."

# Demo
twin_data = {'temperature': 45, 'status': 'running', 'health': 0.85}
copilot = TwinCopilot(twin_data)

queries = ["What is the temperature?", "What's the system status?", "Any recommendations?"]
for q in queries:
    print(f"Q: {q}")
    print(f"A: {copilot.answer_query(q)}\n")
```
