## What is Business Value and KPIs?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Measuring whether your digital twin investment is paying off.

### Technical Definition

Metrics and methodologies for measuring ROI, efficiency gains, and business impact of digital twin investments.

```
# Business Value Demo
@dataclass
class TwinKPI:
    """Key Performance Indicator for twins."""
    name: str
    baseline: float
    current: float
    target: float
    unit: str
    
    def improvement(self) -> float:
        return (self.baseline - self.current) / self.baseline * 100

kpis = [
    TwinKPI("Downtime", 100, 70, 50, "hours/year"),
    TwinKPI("Maintenance Cost", 100000, 80000, 70000, "$/year"),
    TwinKPI("Energy Usage", 1000, 850, 800, "MWh/year"),
]

print("Digital Twin ROI Analysis")
print("=" * 50)
for kpi in kpis:
    print(f"\n{kpi.name}:")
    print(f"  Baseline: {kpi.baseline} {kpi.unit}")
    print(f"  Current: {kpi.current} {kpi.unit}")
    print(f"  Improvement: {kpi.improvement():.1f}%")
    print(f"  Target: {kpi.target} {kpi.unit}")
```

## Why Business Value and KPIs Matters

Measuring business value justifies investment and guides continuous improvement of digital twin implementations.
