## What is Simulation and Physics Models?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Simulations let you predict the future - like a weather forecast for your equipment.

### Technical Definition

Mathematical and computational models that represent physical behavior, enabling prediction and what-if analysis in digital twins.

```
# Physics Simulation Demo
class ThermalModel:
    """Simple thermal physics model."""
    def __init__(self, thermal_mass: float = 100.0, heat_transfer: float = 0.1):
        self.thermal_mass = thermal_mass
        self.heat_transfer = heat_transfer
    
    def simulate_step(self, current_temp: float, ambient_temp: float, 
                      heat_input: float, dt: float = 1.0) -> float:
        heat_loss = self.heat_transfer * (current_temp - ambient_temp)
        temp_change = (heat_input - heat_loss) / self.thermal_mass * dt
        return current_temp + temp_change

# Run simulation
model = ThermalModel()
temp = 25.0
temps = [temp]

for _ in range(50):
    temp = model.simulate_step(temp, 20.0, 50.0)
    temps.append(temp)

plt.plot(temps)
plt.xlabel('Time')
plt.ylabel('Temperature (°C)')
plt.title('Thermal Simulation')
plt.show()
```

## Why Simulation and Physics Models Matters

Simulations enable predictive capabilities - the ability to forecast failures, test scenarios, and optimize operations before making physical changes.
