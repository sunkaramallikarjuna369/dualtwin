@dataclass
class WaterTank:
    tank_id: str
    capacity: float = 1000.0
    current_level: float = 500.0
    
    def fill(self, amount: float) -> None:
        self.current_level = min(self.capacity, self.current_level + amount)
    
    def drain(self, amount: float) -> None:
        self.current_level = max(0, self.current_level - amount)
    
    def read_level(self) -> float:
        return self.current_level + random.gauss(0, 5)


@dataclass
class WaterTankTwin:
    twin_id: str
    tank_id: str
    current_level: float = 0.0
    history: List[Dict] = field(default_factory=list)
    
    def synchronize(self, level_reading: float) -> None:
        self.current_level = level_reading
        self.history.append({'timestamp': datetime.now(), 'level': level_reading})
    
    def get_fill_percentage(self) -> float:
        return (self.current_level / 1000.0) * 100
