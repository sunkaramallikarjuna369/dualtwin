## What is Twin Security and Access Control?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Protecting your digital twin like you would protect your home - with locks, alarms, and access controls.

### Technical Definition

Security frameworks including authentication, authorization, encryption, and audit logging for digital twin systems.

```
# Security Demo
import hashlib
from enum import Enum

class AccessLevel(Enum):
    READ = "read"
    WRITE = "write"
    ADMIN = "admin"

@dataclass
class User:
    user_id: str
    access_level: AccessLevel
    
@dataclass
class SecureTwin:
    """Twin with access control."""
    twin_id: str
    data: Dict = field(default_factory=dict)
    audit_log: List[Dict] = field(default_factory=list)
    
    def access(self, user: User, operation: str) -> bool:
        allowed = False
        if operation == "read" and user.access_level in [AccessLevel.READ, AccessLevel.WRITE, AccessLevel.ADMIN]:
            allowed = True
        elif operation == "write" and user.access_level in [AccessLevel.WRITE, AccessLevel.ADMIN]:
            allowed = True
        
        self.audit_log.append({
            'user': user.user_id,
            'operation': operation,
            'allowed': allowed,
            'timestamp': datetime.now()
        })
        return allowed

# Demo
twin = SecureTwin("DT-001", data={'temp': 25})
reader = User("user1", AccessLevel.READ)
writer = User("user2", AccessLevel.WRITE)

print(f"Reader can read: {twin.access(reader, 'read')}")
print(f"Reader can write: {twin.access(reader, 'write')}")
print(f"Writer can write: {twin.access(writer, 'write')}")
```

## Why Twin Security and Access Control Matters

Security is critical because digital twins often contain sensitive operational data and can potentially control physical systems.
