## What is Dual-Twin Architecture?

This section explains the core concepts in both non-technical and technical terms.

### Non-Technical Explanation

Imagine having two mirrors instead of one - one showing a single machine, another showing the entire production line. Together they give you a complete picture.

### Technical Definition

A design pattern where multiple digital twins are interconnected to provide complementary perspectives, typically pairing asset twins with process twins.

```
# Dual-Twin Architecture Demo
@dataclass
class AssetTwin:
    """Individual equipment twin."""
    twin_id: str
    name: str
    health: float = 1.0
    output_rate: float = 100.0

@dataclass
class ProcessTwin:
    """Process-level twin that aggregates assets."""
    twin_id: str
    assets: List[AssetTwin] = field(default_factory=list)
    
    def get_throughput(self) -> float:
        if not self.assets:
            return 0
        return min(a.output_rate * a.health for a in self.assets)

# Create dual-twin structure
pump = AssetTwin("DT-PUMP", "Pump-001", health=0.9)
motor = AssetTwin("DT-MOTOR", "Motor-001", health=1.0)
process = ProcessTwin("DT-PROCESS", assets=[pump, motor])

print(f"Process throughput: {process.get_throughput():.1f}%")
print(f"Bottleneck: {min(process.assets, key=lambda a: a.health).name}")
```

## Why Dual-Twin Architecture Matters

Dual-twin architecture enables holistic visibility, root cause analysis across system levels, and optimization that considers both equipment and process perspectives.
